# Run a block

Launches the programs of a block in separate windows, in the right order: markers, EEG, decoder, saver.
Before running: open LabRecorder if saving; start the amplifier software if the EEG is not simulated.
Participant, session and task of the saved file are set in `bci/config.py`.

```python
%load_ext autoreload
%autoreload 2

from bci import config
from bci.launcher import launch, stop

EEG_SIMULATED = True
SAVE = True
MODEL_PATH = None # .pkl model file to load; if None, no decoder is used (calibration only)

processes = launch(EEG_SIMULATED, MODEL_PATH, SAVE)
```

```python
stop(processes)
```
